```python
menu = {
    "pizza": 250,
    "burger":150,
    "coffee":80,
    "pasta":200
}

orders=[]

def view_menu():
    print( "\n -----menu-----")
    for i in menu:
        print(i,"RS",menu[i])

def order_food():
   i = input("enter the order")
   if i in menu:
    quantity = int(input("enter the quantity"))
    orders.append([i,quantity])
    print("item added successfully")
   else:
    print("item not found")


def view_bill():
    if len(orders)==0:
        print("no orders found")
        return
    
    total_bill = 0

       
    for order in orders:
         i = order[0]
         quantity = order[1]
         price = quantity * menu[i]
         total_bill += price

         print(i,"x",quantity,"=",price)
         
         print("\n----------------------")
         print("total amount:",total_bill)

while True:
   print("\n ------menu-----") 
   print("1. view_menu")
   print("2. order_food") 
   print("3. view_bill")  
   print("4.exit")

   choice=int(input("enter your choice:"))

   if choice == 1 :
      view_menu()
   elif choice == 2:
      order_food()
   elif choice == 3:
      view_bill()
   elif choice == 4:
      print("thank you")
      break

else:
      print("choice not found")
```
